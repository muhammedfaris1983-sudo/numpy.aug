```python
import numpy as np
# create a 4x4 matrix
z=np.array([[1,2,3,4],[3,2,1,5],[9,8,7,6],[4,5,3,2]])
print(z)
print(z.shape)
print(z.size)
print(z.ndim)

print(z[0]) # first row print chayan vendi
print(z[2]) #
print(z[3])
print(z[1][1])# centre olla colum value print chayan
print(z[1,1]) # centre olla colum value print chayan
print(z[0:,1:3])  # 2 and 3 colum value print chayan
print(z[2,1:3]) # 3rd columnthnu  1,2 index print chayan
print(z[1:3,1:3])
```

```python
#zero matrix
z=np.zeros([2,2],dtype=int) # int kodukupol float mari integer agum
print(z)
```

```python
#one matrix
z=np.ones([3,3],dtype=complex)
print(z)
```

```python
#create two matrix and store it in two variables
a=np.array([[1,2],[3,4]])
print(a)
b=np.array([[5,6],[7,8]])
print(b)
print(a+b) # matrix thammill add chayan
print(a-b) # matrix thammill mins chayan
print(a*b)# not matrix multiplication
print(a@b) # matrix multiplication chayan @ use chayune
print(a.dot(b)) # matrix multiplication chayan olla function
```

```python
#full matrix
z=np.full([3,3],4)
print(z)
```

```python
#create a 4x4 matrix  with value 9
z=np.full([4,4],9)
print(z)
```

```python
#identity matrix
z=np.identity(3)
print(z)
z=np.eye(4,dtype=int) # eye use chaythum identity matrix  kandpidikam
print(z)
```

```python
#create 3x4 matrix
z=np.array([[3.1,4.5,7.7,8.9],[2.9,6.3,5.5,8.5],[1.6,2.7,3.7,8.2]])
print(z)
print(np.floor(z))
print(np.ceil(z))
print(np.round(z)) # nearest even
```

```python
z=np.array([[3.1,4.5,7.7,8.9],[2.9,6.3,5.5,8.5],[1.6,2.7,3.7,8.2]])
print(z.max()) #max kandupidikan
print(z.min())# minum kandupidikan
print(z.sum()) # sum kandupidikan
```

```python
z=np.array([[3.1,4.5,7.7,8.9],[2.9,6.3,5.5,8.5],[1.6,2.7,3.7,8.2]])
print(z.max(axis=1)) # rownte max value kandupidika vendi axis =1
print(z.max(axis=0))# columnte max value kandupidikan vendi axis =0

print(np.sqrt(z)) # square root kandupidikan sqrt use chayunu
print(np.sin(z)) # sin kandupidikan sin use chayan
print(np.cos(z)) # cos kandupidikan use chayan

```
